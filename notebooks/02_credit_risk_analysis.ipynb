```python
import pandas as pd

credit = pd.read_excel(
    "../data/default of credit card clients.xls",
    header=1
)

print("Dataset dimensions:", credit.shape)
print("\nColumn names:")
print(credit.columns.tolist())

display(credit.head())
```

```python
# Select 500 customers randomly
credit_sample = credit.sample(
    n=500,
    random_state=42
)

# Reset row numbering
credit_sample = credit_sample.reset_index(drop=True)

print("Sample dimensions:", credit_sample.shape)

display(credit_sample.head())
```

```python
# Count customers in each default category
default_counts = credit_sample[
    "default payment next month"
].value_counts()

print("Default distribution:")
print(default_counts)

print("\nPercentage distribution:")
print(credit_sample["default payment next month"].value_counts(normalize=True) * 100)
```

```python
# Select the variables used to predict default

X = credit_sample[
    ["LIMIT_BAL", "AGE", "PAY_0", "BILL_AMT1", "PAY_AMT1"]
]

# Select the outcome we want to predict
y = credit_sample["default payment next month"]

print("Input variables:")
display(X.head())

print("Target variable:")
display(y.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# Create the model
model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

# Train the model
model.fit(X_train, y_train)

print("Logistic regression model trained successfully!")
```

```python
# Predict default outcomes
y_pred = model.predict(X_test)

# Predict default probabilities
y_prob = model.predict_proba(X_test)[:, 1]

print("Actual outcomes:")
print(y_test.head(10).tolist())

print("\nPredicted outcomes:")
print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

# Calculate confusion matrix
cm = confusion_matrix(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_auc_score

auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC Score:", round(auc, 3))
```

```python
# Save credit-risk model results

from sklearn.metrics import precision_score, recall_score, f1_score

credit_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "ROC-AUC",
        "Default Precision",
        "Default Recall",
        "Default F1-score"
    ],
    "Value": [
        accuracy,
        auc,
        precision_score(y_test, y_pred),
        recall_score(y_test, y_pred),
        f1_score(y_test, y_pred)
    ]
})

display(credit_results.round(3))

credit_results.to_csv(
    "../outputs/credit_risk_results.csv",
    index=False
)

print("Credit-risk results saved successfully!")
```

# Credit Risk Analysis: Credit Card Default Prediction

## Objective
The objective of this analysis was to examine credit default behaviour and build a basic statistical model to estimate the likelihood of customer default.

## Dataset
The analysis uses the Default of Credit Card Clients dataset from the UCI Machine Learning Repository.

A random sample of 500 customers was selected from the original 30,000 observations.

## Methodology
- Selected five explanatory variables: credit limit, age, repayment status, bill amount and payment amount.
- Divided the sample into 80% training data and 20% testing data.
- Applied feature standardisation.
- Developed a logistic regression model to predict customer default.
- Evaluated the model using accuracy, precision, recall, F1-score and ROC-AUC.

## Model Results

- Accuracy: 79%
- ROC-AUC: 0.721
- Default precision: 67%
- Default recall: 25%
- Default F1-score: 36%

## Interpretation
The model achieved 79% accuracy and a ROC-AUC score of 0.721, indicating some ability to distinguish between defaulters and non-defaulters.

However, default recall was only 25%. This means the model identified only 6 of the 24 actual defaulters in the test sample.

The results demonstrate why accuracy alone is insufficient when evaluating credit-risk models. Identifying potential defaulters is particularly important in lending decisions.

## Limitations
- The model uses only 500 observations.
- Only five explanatory variables were included.
- The dataset represents Taiwanese credit-card customers from 2005.
- The model is intended for educational analysis and is not suitable for actual lending decisions.

## Conclusion
This exercise demonstrates how logistic regression can be applied to credit-risk analysis and how model performance can be evaluated using multiple classification metrics.
